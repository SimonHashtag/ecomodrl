# Portfolio Choice Model with Buffer Stock Savings

### Preferences, income, normalization

$$
\max_{\{c_t,\omega_t\}} \ \mathbb{E}_0\sum_{t=0}^\infty \beta^t \frac{c_t^{1-\rho}}{1-\rho},
\quad \rho>0,\ \beta\in(0,1).
$$

* Permanent income: $P_{t+1}=G\Psi_{t+1}P_t$, so $\Gamma_{t+1}\equiv P_{t+1}/P_t=G\Psi_{t+1}$.

* Transitory income: $Y_{t+1}=P_{t+1}\Theta_{t+1}$.

* Normalize by $P_t$: $m_t=M_t/P_t,\ c_t=C_t/P_t$.

### Assets and states

* One **risk-free** asset with constant gross return $R_f>0$.
* $N$ **risky** assets with **lognormal** gross returns whose **log-means** depend on own predictors:
  $$
  z_{j,t+1}=\rho_j z_{j,t}+\sigma_{z,j}\varepsilon_{j,t+1},\quad |\rho_j|<1,
  $$
  $$
  \ln R_{j,t+1}=\kappa_{j}+\xi_j z_{j,t}+\sigma_{j}\eta_{j,t+1},\qquad j=1,\dots,N
  $$
  $\varepsilon,\eta$ are i.i.d. standardized shocks (correlations optional). With this specification, **$R_{j,t+1}>0$** always.

* Portfolio weights on $(N+1)$ assets (risk-free indexed by $0$): $\omega_t=(\omega_{0,t},\dots,\omega_{N,t})\in\Delta^{N+1}=\{\omega\ge 0,\ \sum_i\omega_i=1\}$.

* Portfolio gross return: $R^p_{t+1}=\sum_{i=0}^N \omega_{i,t} R_{i,t+1}$ with $R_{0,t+1}=R_f$.

### Budget and Bellman equation

$$
m_{t+1}=\frac{R^p_{t+1}}{\Gamma_{t+1}}a_t+\Theta_{t+1},\qquad a_t=m_t-c_t\ge\underline a \quad \text{(Borrowing limit)}
$$
$$
V(m,\mathbf z)=\max_{\substack{c\in[0,m],\omega\in\Delta^{N+1}}}\Big\{u(c)+\beta\mathbb{E}[V(m',\mathbf z')]\Big\}.
$$

First-order condition (Euler) when borrowing limit constraint is slack:
$$
u'(c_t)=\beta\mathbb{E}_t\left[\frac{R^p_{t+1}}{\Gamma_{t+1}}u'(c_{t+1})\right]. \tag{E}
$$

### Portfolio FOCs: deriving the **common multiplier $\lambda_t$** and the **orthogonality (Eq. 4)**

**Date-$t$** Lagrangian:
$$
\mathcal{L}_t \equiv u(c_t)+\beta\mathbb{E}_t\left[V(m_{t+1},\mathbf z_{t+1})\right]+\nu_t\left(1-\sum_{i=0}^N\omega_{i,t}\right)+\sum_{i=0}^N \mu_{i,t},\omega_{i,t},
$$
with multipliers $\nu_t\in\mathbb{R}$ (simplex sum) and $\mu_{i,t}\ge 0$ (no-short constraints $\omega_{i,t}\ge 0$).

Because $m_{t+1}=\sum_i \omega_{i,t} R_{i,t+1} \frac{a_t}{\Gamma_{t+1}} +\Theta_{t+1}$,
$$
\frac{\partial m_{t+1}}{\partial \omega_{i,t}}=R_{i,t+1} \frac{a_t}{\Gamma_{t+1}}.
$$
By the envelope condition at $t+1$, $\partial V(m_{t+1},\mathbf z_{t+1})/\partial m_{t+1}=u'(c_{t+1})$ (one more unit of next-period cash-on-hand can be consumed next period). Hence,
$$
\frac{\partial}{\partial \omega_{i,t}}\beta\mathbb{E}_t[V(m_{t+1},\mathbf z_{t+1})]
=\beta\mathbb{E}_t\Big\{u'(c_{t+1})\frac{a_t}{\Gamma_{t+1}}R_{i,t+1}\Big\}.
$$
FOC for $\omega_{i,t}$:
$$
\beta\mathbb{E}_t\Big[u'(c_{t+1})\frac{a_t}{\Gamma_{t+1}}R_{i,t+1}\Big]-\nu_t+\mu_{i,t}=0.
$$
For **interior** assets $(\omega_{i,t}>0\Rightarrow \mu_{i,t}=0)$:
$$
\beta\mathbb{E}_t\Big[u'(c_{t+1})\frac{a_t}{\Gamma_{t+1}}R_{i,t+1}\Big]=\nu_t\qquad\forall i \text{ with }\omega_{i,t}>0.
$$
Divide both sides by $a_t u'(c_t)$ and **define the normalized stochastic discount factor (SDF)/marginal rate of substitution normalized by permanent-income growth**
$$
\mathfrak m_{t+1}\equiv \beta\frac{u'(c_{t+1})}{u'(c_t)}\cdot\frac{1}{\Gamma_{t+1}},\qquad
\lambda_t\equiv\frac{\nu_t}{a_t,u'(c_t)}.
$$
Then for all interior assets
$$
\boxed{\ \mathbb{E}_t\big[\mathfrak m_{t+1}R_{i,t+1}\big]=\lambda_t\ } \qquad (i=0,\dots,N \text{ with }\omega_{i,t}>0).
$$
Because $\sum_i \omega_{i,t}=1$ and $R^p_{t+1}=\sum_i \omega_{i,t}R_{i,t+1}$, we also have
$$
\mathbb{E}_t\big[\mathfrak m_{t+1}R^p_{t+1}\big]=\lambda_t.
$$
Subtracting gives the **pricing orthogonality conditions**:
$$
\boxed{\ \mathbb{E}_t\big[\mathfrak m_{t+1}(R_{i,t+1}-R^p_{t+1})\big]=0\quad\text{if }\omega_{i,t}>0\quad \le 0\ \text{if }\omega_{i,t}=0.\ } \tag{4}
$$
That is, the SDF-weighted excess return of any **actively held** asset has zero conditional mean; at a corner (no-short constraint binding), the mean is non-positive. For a deeper dive see: [LARS PETER HANSEN, JOHN HEATON, JUNGHOON LEE and NIKOLAI ROUSSANOV (2007) - Handbook of Econometrics, Volume 6A, pp. 3976 ff.](https://larspeterhansen.org/wp-content/uploads/2016/10/Intertemporal-Substitution-and-Risk-Aversion.pdf)

# Time iteration

**Idea.** Keep approximations of the **current-period policies** $(c,\omega)$. For each state node $(m,\mathbf z)$, **solve the static first-order conditions** (Euler + portfolio KKT) given your *current* approximation of **next-period** policies (to evaluate $u'(c_{t+1})$ at random $m_{t+1}$). Update the policies; repeat until convergence.

## System to solve (when OBC slack)

Let $a=m-c$. For any candidate $(c,\omega)$:

* Draw shocks $\Xi=(\Psi,\Theta,\eta_{1:N},\varepsilon_{1:N})$; compute
  $\Gamma'=G\Psi$, $R_i'(\mathbf z,\eta)$, $R^{p'}=\sum_i \omega_i R_i'$,
  $m'=\dfrac{R^{p'}}{\Gamma'}a+\Theta$, $\mathbf z'=\Phi(\mathbf z,\varepsilon)$.
* Evaluate $c'(m',\mathbf z')$ from the **current** policy approximation and form $u'(c')$.

Define
$$
g_i(c,\omega \mid m,\mathbf z)=\mathbb{E}\left[\frac{u'(c')}{\Gamma'}R_i'\right],\quad i=0,\dots,N.
$$
Then the **Euler** and **portfolio** conditions become (interior case):
$$
F_0(c,\omega)\equiv u'(c)-\beta\sum_{i=0}^N \omega_i g_i = 0,
$$
$$
F_i(c,\omega)\equiv g_i-\sum_{j=0}^N \omega_j g_j = 0\quad(\text{for each active }i). 
$$

The simplex $\sum\omega_i=1$ and bounds $\omega_i\ge 0$ also apply.

### Corners

After a solve, if any $\omega_i<\varepsilon$, set $\omega_i=0$, **drop** its KKT condition, and re-solve on that **active set** (smaller face of the simplex). This “active-set” TI is very fast with **warm starts** (use last iteration’s $\omega$).

### Practical parameterization for the solve

* Enforce the simplex via **softmax**: $\omega_i=\exp(\theta_i)/\sum_j\exp(\theta_j)$.
* Keep $c$ in bounds with, e.g., $c=\sigma(\chi)\cdot(m-\underline a)$ where $\sigma$ is a logistic.
* Then solve the **unconstrained** root problem in $(\chi,\theta_0,\dots,\theta_N)$ for the equations.

## Full TI algorithm

**Inputs.** Grids for $m$ and $\mathbf z$; quadrature/MC nodes for $\Xi$; tolerances.

**Initialize.**

* $c^{(0)}(m,\mathbf z)$: e.g., a buffer-stock policy from the one-asset model or a fraction of $m$.
* $\omega^{(0)}(m,\mathbf z)$: e.g., all in $R_f$ or a myopic share.

**Iterate $k=0,1,2,\dots$:** (can parallelize across grid nodes)

For each node $(m,\mathbf z)$:

1. **Warm start.** Take $(c,\omega)\leftarrow(c^{(k)}(m,\mathbf z),\omega^{(k)}(m,\mathbf z))$.

2. **(Local) solve the static system.**

   Using the expectations defined above (with **common random numbers** across iterations), solve:

   * **Euler:** $u'(c)-\beta\sum_i \omega_i g_i=0$.
   * **Portfolio equalization** on the current **active set**: $g_i=\sum_j\omega_j g_j$ for active (i).
   * **Simplex:** via softmax parameterization.
     Use a small **Newton / quasi-Newton with line search** (unknowns = $1+\#\text{active}$). If some $\omega_i\to 0$, drop it and re-solve.
     Special case $N=1$: it’s just **2 unknowns** $(c,\alpha)$ with **2 equations** (Euler + $g_1=g_0$); a scalar root in $\alpha$ nested inside a 1D Newton in (c) is extremely fast.

3. **Record update.** Set $c^{(k+1)}(m,\mathbf z)\leftarrow c^\star$, $\omega^{(k+1)}(m,\mathbf z)\leftarrow\omega^\star$.

4. **Borrowing constraint (OBC) splice.**
   If the interior solve tries to use $a=m-c<\underline a$, set **binding** $a=\underline a\Rightarrow c=m-\underline a$.

   * Euler becomes an **inequality** (LHS $\ge$ RHS) automatically at the solution.
   * Portfolio step: re-solve KKT with this fixed (a) (same active-set logic).

**Convergence.** Stop when $\|c^{(k+1)}-c^{(k)}\|_\infty$ and $\|\omega^{(k+1)}-\omega^{(k)}\|_\infty$ fall below tolerance (e.g., $10^{-6}$). 

Howard acceleration (extra policy improvement steps without recomputing expectations) is possible but usually unnecessary if you warm-start well.

## Two quick specializations

* **Two assets (safe + one risky).** Unknowns: $c\in(0,m-\underline a]$, $\alpha\in[0,1]$. Solve:
  $$
  \begin{cases}
  u'(c) - \beta\mathbb{E}[(R^p/\Gamma')u'(c')] = 0,\
  \mathbb{E}\left[\dfrac{u'(c')}{\Gamma'}(R_e'-R_f)\right]=0\quad(\text{if }0<\alpha<1),
  \end{cases}
  $$
  with corner inequalities for $\alpha=0,1$. This is **very fast** with a nested scalar root in $\alpha$ inside a 1D Newton in $c$.

* **Many assets with likely corners.** Use **softmax**; solve equalizations on the current active set; keep a tiny $\varepsilon$ floor to stabilize (e.g., $10^{-10}$) and project to zero when below.

## Nested-EGM

State: $(m,\mathbf z)$. Controls: $c\in[0,m]$, $a'=m-c\ge\underline a$, and $\omega\in\Delta^{N+1}$.

**Step 0 — Grids and policies.**
Choose grids $ \mathcal G_{a}$ and $ \mathcal G_{\mathbf z}$. Maintain policy approximations $c^{(k)}(m,\mathbf z)$ and $\omega^{(k)}(a,\mathbf z)$. Fix quadrature / quasi-MC nodes for shocks $\Xi=(\Psi,\Theta,\eta_{1:N},\varepsilon_{1:N})$.

**At each node $(a,\mathbf z)$:**

1. **Build conditional objects per shock draw.**

   Given $\mathbf z$ and a, draw
   $$
   \Gamma'=G\Psi,\quad \ln R_{j}'=\mu_{j}+\lambda_j z_j + \sigma_{j}\eta_{j},\quad R_0'=R_f,
   $$
   then
   $$
   R^{p'}(\omega)=\sum_{i=0}^N \omega_i^{(k)}(a, \mathbf z) R_i',\qquad
   m'=\frac{R^{p'}(\omega)}{\Gamma'}a + \Theta',\qquad
   \mathbf z'=\Phi(\mathbf z,\varepsilon).
   $$
   Evaluate **continuation consumption** $c_{t+1}=c^{(k)}(m',\mathbf z')$ and $u'(c_{t+1})$.

2. **Form asset-wise expected marginal-utility carriers.**

   For each asset $i=0,\dots,N$,
   $$
   g_i(a,\mathbf z;\omega)\equiv
   \mathbb E\Big[ \tfrac{u'(c_{t+1})}{\Gamma'}R_i' \Big| a,\mathbf z,\omega \Big].
   $$
   Note $g_i$ **depends on $\omega$** both directly (through $R_i'$) and indirectly via $m'$ inside $u'(c_{t+1})$.

3. **Solve the portfolio KKT at $(a,\mathbf z)$.**

   Let $\bar g(\omega)=\sum_{i=0}^N \omega_i g_i(a,\mathbf z;\omega)$.
   
   The **Euler equation** is
   $$
   u'(c_t)=\beta\bar g(\omega),
   $$
   and the **portfolio FOCs/KKT** (no-short, simplex) are
   $$
   \mathbb E\big[\mathfrak m_{t+1} R_i'\big]=\lambda_t \ \ \Longleftrightarrow\ \
   \frac{\beta}{u'(c_t)}g_i(a,\mathbf z;\omega)=\lambda_t,
   $$
   so for any **interior** asset the $g_i$’s are **equalized**. Practically, solve
   $$
   \omega^{(k+1)}(a,\mathbf z)\in\arg\min_{\omega\in\Delta^{N+1}}
   \sum_{i=0}^N\Big(g_i(a,\mathbf z;\omega)-\bar g(\omega)\Big)_+^2
   $$
   by a projected Newton/interior-point method with **warm start** $\omega^{(k)}$. (Complementarity penalties handle bound assets.)

> This yields $\omega^{(k+1)}$, the **common multiplier** follows as
> $$
> \lambda_t = \frac{\beta}{u'(c_t)}g_i(a,\mathbf z;\omega^{(k+1)}),
> $$
> equal for all interior (i). Subtracting $\sum_j\omega_j^{(k+1)} (\cdot)$ gives the **orthogonality**:
> $$
> \mathbb E\big[\mathfrak m_{t+1}(R_i'-R^{p'})\big]=0\ \ \text{if}\ \omega_i^{(k+1)}>0,\quad \le 0\ \text{if}\ \omega_i^{(k+1)}=0.
> $$

4. **Invert the scalar Euler (EGM step).**
   With $\omega^{(k+1)}$ fixed, compute $\bar g^\star=\bar g(\omega^{(k+1)})$ and set
   $$
   u'(c^{\text{pre}})=\beta\bar g^\star
   \quad\Rightarrow\quad
   c^{\text{pre}}(a,\mathbf z)=\big(u'\big)^{-1}\big(\beta,\bar g^\star\big).
   $$

5. **Map to the current-period grid.**
   The endogenous current cash-on-hand is simply
   $$
   m^{\text{pre}}(a,\mathbf z)=c^{\text{pre}}(a,\mathbf z)+a.
   $$
   Record the triplet $\big(m^{\text{pre}},c^{\text{pre}},\omega^{(k+1)}\big)$.

**Post-processing (per $\mathbf z$):** sort by $m^{\text{pre}}$, build **monotone** interpolants for $c^{(k+1)}(m,\mathbf z)$; splice in the **OBC branch** for low $m$ (set $a'=\underline a\Rightarrow c=m-\underline a$, and solve the small KKT if you want portfolio at the boundary). Iterate $k\gets k{+}1$ to convergence. 

After convergence also determine $\omega(m,\mathbf z)$

## 4.4 Notes on practicality and sources

* With one predictor per asset (state dimension $N+1$), expectations become the bottleneck; sparse quadrature / low-discrepancy MC and common random numbers are essential. General pros/cons and scaling issues are surveyed in recent computational reviews. 
* **EGM** (consumption-saving margin) is the core acceleration (Carroll, 2006). **NEGM / generalized EGM** describes the practice of combining EGM with local solves for extra margins or non-convexities (Druedahl & Jørgensen, 2017; Druedahl, 2021).
