# Derivation of the Euler Equation with Permanent–Transitory Income

## 1. Primitives

- **Preferences**

  The household chooses $\{C_t\}_{t\ge 0}$ to maximize expected lifetime utility
  $$
  \max_{\{C_t\}_{t=0}^{\infty}} \; E_0\left[\sum_{t=0}^\infty \beta^t u(C_t)\right],
  \qquad 0<\beta<1.
  $$

- **Asset accumulation**

  One asset with gross return $R>0$:
  $$
  K_{t+1} = R\big(K_t + Y_t - C_t\big).
  $$

- **Income process**

  Income is
  $$
  Y_t = P_t\,\Theta_t,
  $$
  where:
  - Permanent component:
    $$
    P_{t+1} = G\,P_t\,\Psi_{t+1}, \qquad
    \ln \Psi_{t+1} \sim \mathcal N\left(-\tfrac{1}{2}\sigma_\Psi^2,\sigma_\Psi^2\right),
    \quad E[\Psi_{t+1}]=1.
    $$

  - Transitory component:
    $$
    \Theta_t = 
    \begin{cases}
    \underline{\Theta} & \text{with prob } p, \\[2mm]
    Z_t & \text{with prob } 1-p,
    \end{cases}
    \qquad \ln Z_t \sim \mathcal N(\mu_Z,\sigma_Z^2),
    \quad E[\Theta]=1.
    $$

---

## 2. Cash-on-hand in levels

Define
$$
M_t \equiv K_t + Y_t.
$$

Then
$$
K_{t+1} = R\big(K_t + Y_t - C_t\big) = R(M_t - C_t),
$$
$$
Y_{t+1} = P_{t+1}\Theta_{t+1} = G\,P_t\,\Psi_{t+1}\,\Theta_{t+1}.
$$

So
$$
M_{t+1} = K_{t+1} + Y_{t+1}
= R(M_t - C_t) + G\,P_t\,\Psi_{t+1}\,\Theta_{t+1}.
$$

---

## 3. Normalization by permanent income

Define normalized variables
$$
m_t = \frac{M_t}{P_t}, \qquad c_t = \frac{C_t}{P_t}.
$$

Divide by $P_{t+1} = G\,P_t\,\Psi_{t+1}$:
$$
\frac{M_{t+1}}{P_{t+1}}
= \frac{R}{G\,\Psi_{t+1}}\frac{M_t - C_t}{P_t} + \Theta_{t+1}
= \frac{R}{G\,\Psi_{t+1}}(m_t - c_t) + \Theta_{t+1}.
$$

Hence
$$
\boxed{m_{t+1} = \frac{R}{G\,\Psi_{t+1}}\,(m_t - c_t) + \Theta_{t+1}}.
$$

This equation describes the evolution of normalized resources.

---

## 4. Lagrangian

Introduce multipliers $\{\lambda_{t+1}\}$ for the constraints.  
After normalizing consumption by $P_t$, per-period utility becomes $P_t^{\,1-\gamma}u(c_t)$ under CRRA ($u(C_t)=P_t^{\,1-\gamma}u(c_t)$). The Lagrangian is
$$
\mathcal L
= E_0\sum_{t=0}^\infty \beta^t \Big\{
P_t^{\,1-\gamma}u(c_t) 
+ \lambda_{t+1}\big[m_{t+1} - \tfrac{R}{G\Psi_{t+1}}(m_t - c_t) - \Theta_{t+1}\big]
\Big\}.
$$

---

## 5. First-order conditions

### With respect to $c_t$:
$$
\frac{\partial \mathcal L}{\partial c_t}:
\quad \beta^t\Big(P_t^{\,1-\gamma} u'(c_t) + E_t\!\big[\lambda_{t+1}\tfrac{R}{G\Psi_{t+1}}\big]\Big)=0
\quad \Rightarrow \quad
P_t^{\,1-\gamma} u'(c_t) = -\,E_t\!\left[\lambda_{t+1}\frac{R}{G\Psi_{t+1}}\right].
\tag{A}
$$

### With respect to $m_t$:
$m_t$ appears in the $t-1$ and $t$ constraints, yielding
$$
\frac{\partial \mathcal L}{\partial m_t}:
\quad \beta^{t-1}\lambda_t - \beta^{t}E_t\!\big[\lambda_{t+1}\tfrac{R}{G\Psi_{t+1}}\big]=0
\quad \Rightarrow \quad
\lambda_t = \beta\,E_t\!\left[\lambda_{t+1}\frac{R}{G\Psi_{t+1}}\right].
\tag{B}
$$

---

## 6. Combining (A) and (B)

Shift (A) forward one period:
$$
P_{t+1}^{\,1-\gamma} u'(c_{t+1})
= -\,E_{t+1}\!\left[\lambda_{t+2}\frac{R}{G\Psi_{t+2}}\right].
$$

Use (B) at $t+1$ to eliminate the expectation:
$$
\lambda_{t+1}
= \beta\,E_{t+1}\!\left[\lambda_{t+2}\frac{R}{G\Psi_{t+2}}\right]
= -\,\beta\,P_{t+1}^{\,1-\gamma} u'(c_{t+1}).
$$

Substitute this into (A) at time $t$:
$$
P_t^{\,1-\gamma}u'(c_t)
= \beta\,E_t\!\left[P_{t+1}^{\,1-\gamma}\,u'(c_{t+1})\,\frac{R}{G\Psi_{t+1}}\right].
$$

Since $P_{t+1}^{\,1-\gamma}=(G\Psi_{t+1})^{\,1-\gamma}P_t^{\,1-\gamma}$, cancel $P_t^{\,1-\gamma}$ to obtain
$$
\boxed{\,u'(c_t) = \beta R \, E_t\!\left[(G\Psi_{t+1})^{-\gamma}\,u'(c_{t+1})\right].\,}
$$

---

## 7. Economic intuition

- The **LHS**: marginal utility of consuming one unit today (in units scaled by $P_t$).
- The **RHS**: discounted expected marginal utility of saving that unit. Savings grow at return $R$, but their value next period is scaled down by growth in permanent income $G\Psi_{t+1}$; under CRRA this yields the factor $(G\Psi_{t+1})^{-\gamma}$.
- A larger permanent shock $\Psi_{t+1}$ reduces the marginal value of assets tomorrow, tilting optimal consumption growth.

Together with the transition equation
$$
m_{t+1} = \frac{R}{G\Psi_{t+1}}(m_t - c_t) + \Theta_{t+1},
$$
this Euler condition fully characterizes the household’s intertemporal consumption choice in the normalized model.

---

## 8. Endogenous Grid Method (EGM)

### Objects and grid

- Define resources saved before shocks:
  $$
  rs_t \equiv m_t - c_t = (1-a_t)m_t.
  $$

- Given a current guess for the policy $a^{\text{old}}(m)$, the goal is to construct an updated policy $a^{\text{new}}(m)$.

### One EGM pass (mapping $a^{\text{old}}\mapsto a^{\text{new}}$)

For each grid point:

1. Next-period cash-on-hand as a function of shocks $(\Theta,\Psi)$  
   Using the normalized transition,
   $$
   m_{t+1}(rs;\Theta,\Psi) \;=\; \frac{R}{G\,\Psi}\, rs \;+\; \Theta .
   $$

2. Continuation consumption (using the continuation policy)  
   $$
   c_{t+1}(rs;\Theta,\Psi) \;=\; a^{\text{old}}\!\big(m_{t+1}(rs;\Theta,\Psi)\big)\cdot m_{t+1}(rs;\Theta,\Psi).
   $$

3. Back out current consumption from the Euler equation  
   With CRRA and permanent-income normalization, the Euler condition is
   $$
   u'(c_t) \;=\; \beta R \; E_t\!\left[(G\Psi_{t+1})^{-\gamma} u'\!\left(c_{t+1}\right)\right]
   \;=\; \frac{\beta R}{G^\gamma}\; E_t\!\left[\Psi_{t+1}^{-\gamma} u'\!\left(c_{t+1}\right)\right].
   $$
   Denote
   $$
   \mu(rs) \;\equiv\; \frac{\beta R}{G^\gamma}\; E\!\left[\Psi^{-\gamma} u'\!\big(c_{t+1}(rs;\Theta,\Psi)\big)\right].
   $$
   Quadrature: approximate the expectations with Gauss–Hermite nodes/weights for  $\Psi$ and $\Theta$. Form joint weights by the product of the respective marginal weights. This yields
   $$
   \mu(rs) \;=\; \frac{\beta R}{G^\gamma}\sum_{j,k}\omega_{jk}\,\Psi_k^{-\gamma}\,u'\!\big(c_{t+1}(rs;\Theta_j,\Psi_k)\big).
   $$

   Then invert marginal utility to obtain
   $$
   c_t(rs) \;=\; u'^{-1}\!\big(\mu(rs)\big).
   $$

4. Map to the endogenous current grid for resources
   $$
   m_t(rs) \;=\; c_t(rs) \;+\; rs.
   $$

5. Implied current-period consumption share
   $$
   a_t(rs) \;=\; \frac{c_t(rs)}{m_t(rs)}.
   $$

This yields a set of endogenous pairs $\{(m_t(rs^i),\, a_t(rs^i))\}_{i=1}^I$.

### Interpolation and boundary

- Sort the pairs by $m_t(rs^i)$ and build a monotone interpolant $a^{\text{new}}(m)$ over current resources $m$ (e.g., PCHIP).  
- Boundary at zero wealth: enforce $a^{\text{new}}(0)=1$ so that with zero cash-on-hand the household consumes everything.

## Buffer Stock Saving - Carroll (1997)

### The model

We observe a representative household that solves the following intertemporal consumption optimization problem:

$$
    \max_{\left\{C_i\right\}_{i=t}^\infty}\mathbb{E}_t\sum_{i=t}^\infty \beta^{i-t} u(C_i)   
$$
subject to 
$$
    W_{t+1}=R\left[W_t+Y_t-C_t\right] \\
    Y_{t}=P_{t} \theta_{t} \\
    P_{t+1} = G P_{t} \psi_{t+1},    
$$

where $C_t$ is consumption, $\beta=1/(1+\delta)$ is the discount factor, $R=(1+r)$ is the constant gross interest rate and $W_t$ are assets. 

$Y_t$ is income, which is determined by permanent labor income $P_t$ and a transitory component $\theta_{t}$. Permanent income evolves given a growth factor $G=(1+g)$ and a permanent shock $\psi_{t+1}$.

By analysing data from the Panel Study of Income Dynamics, Carrol (1992) finds that a US houshold's income uncertainty is well parameterized by:

$$
\theta \sim 
\left| 
\begin{aligned}
&\underline \theta \quad \text{with probability } p \\
&Z \quad \text{with probability } (1 - p)
\end{aligned}
\right|,
$$

where $$\ln Z \sim TN(\mu_Z, \sigma^2_{\ln Z}),$$

$$
\ln \psi \sim TN\left( \frac{-\sigma^2_{\ln \psi}}{2}, \sigma^2_{\ln \psi} \right)
$$

$TN$ denotes the normal distribution truncated to a range of three standard deviations below and above the mean.

Optimal consumption depends on current gross wealth, i.e. $X_t=W_t+Y_t$, which is lower bounded $X_t\geq \underline X$ and evolves according to

$$
X_{t+1}=R(X_t - C_t) + Y_{t+1}.
$$


### Endogenous grid-point method solution strategy

**Normalize state variable**

To ensure convergence in the infinite horizon setting it is important to have a stationary state variable, so we normalize all variables by permanent income $P_t$ as proposed by Carroll (1996).

Thus, the equation of motion becomes

\begin{align*}
\frac{X_{t+1}}{P_t} &= \frac{R}{P_t}(X_t-C_t) + \frac{Y_{t+1}}{P_t} \\
&= \frac{R}{P_t}(X_t-C_t) + \frac{P_{t+1}\theta_{t+1}}{P_t} \\
&= \frac{R}{P_t}(X_t-C_t) + G\psi_{t+1}\theta_{t+1} \\
\end{align*}

$$
\Rightarrow \frac{X_{t+1}}{G P_t \psi_{t+1}} \equiv x_{t+1} =\frac{R}{G\psi_{t+1}}(x_t-c_t) + \theta_{t+1}
$$

As a side effect we can condense the state space to a single variable $x_t$ instead of the two state variables $W_t$ and $Y_t$.

**Derive Euler equation**

The Lagrangian has the following form now:

$$
\mathcal{L}=\mathbb{E}_t\sum_{i=t}^\infty \left(\beta^{i-t} u(c_i) + \lambda_i (\frac{R}{G\psi_{i+1}}(x_i-c_i) + \theta_{i+1}-x_{i+1}) \right)
$$

The first order conditions are:

$$
\frac{\partial \mathcal{L}}{\partial c_i}=\mathbb{E}_t\left(\beta^{i-t}\frac{\partial u(c_i)}{\partial c_i}-\lambda_i \frac{R}{G\psi_{i+1}}\right) \stackrel{!}{=} 0 \\ 
\Rightarrow \lambda_i = \beta^{i-t} \frac{G}{R} \mathbb{E}_t\left(\frac{\partial u(c_i)}{\partial c_i} \cdot \psi_{i+1}\right)
$$

$$
\frac{\partial \mathcal{L}}{\partial x_{i+1}}=\mathbb{E}_t \left( \lambda_{i+1} \frac{R}{G\psi_{i+2}}-\lambda_i \right) \stackrel{!}{=} 0 \\ 
$$

From this follows the Euler equation

$$
u^\prime(c_i) = \frac{\beta R}{G} \mathbb{E}_t\left(\frac{u^\prime(c_{i+1})}{\psi_{i+1}}\right)
$$

**Endogenous grid-point transformation**

Let us define a consumption policy function $c(x_t)$ that returns a permanent income-normalized consumption amount for each normalized gross wealth. Further, we introduce a post-decision state variable $a_t=x_t-c_t$ so that

$$
x_{t+1} = \frac{R}{G\psi_{t+1}}a_t + \theta_{t+1}
$$

We can rewrite the Euler equation as

\begin{align*}
c &= {u^\prime}^{-1}\left(\frac{\beta R}{G} \mathbb{E}_t\left(\frac{u^\prime(c(x_{i+1}))}{\psi_{i+1}}\right) \right) \\
  &= {u^\prime}^{-1}\left(\frac{\beta R}{G} \mathbb{E}_t\left(\frac{u^\prime(c(\frac{R a}{G\psi_{i+1}} + \theta_{t+1}))}{\psi_{i+1}}\right) \right)
\end{align*}

for given values of the post-decision value $a$. Having calculated $c$ we can recover the pre-decision value by 
$$
x = a + c.
$$
Having determined a set of $(x,c)$ pairs for a given grid of $a$, we can interpolate the set and determine a new consumption policy function $c(x)$. Doing this repetively, the policy converges towards the optimal policy function, i.e. a fixed-point of the Coleman-Refett operator (Coleman 1990, Reffett 1996).

### Further specification of the model

The utility function is defined as $u(c)=\dfrac{c^{1-\rho}-1}{1-\rho}$,   $\rho>0$.

### Excursion: Envelope Theorem

$$
V(X_t, P_t) = \max_{C_t} \left( u(C_t) + \beta V(X_{t+1}, P_{t+1}) \right)
$$

$$
\frac{\partial V(X_t, P_t)}{\partial C_t}
= \frac{\partial u(C_t)}{\partial C_t}
+ \beta \frac{\partial V(X_{t+1}, P_{t+1})}{\partial X_{t+1}} \frac{\partial X_{t+1}}{\partial C_t}
+ \frac{\partial V(X_{t+1}, P_{t+1})}{\partial P_{t+1}} \frac{\partial P_{t+1}}{\partial C_t} = 0
$$

$$
\Rightarrow u'(C_t) = \beta R \frac{\partial V(X_{t+1}, P_{t+1}) }{\partial X_{t+1}} \tag{2}
$$

---

$$
\frac{d V(X_t, P_t)}{d X_t} =
\frac{d u(C_t^*)}{d X_t} +\textcolor{coral}{\frac{\partial u(C_t^*)}{\partial C_t}} \cdot \frac{d C_t^*}{d X_t}
+ \beta \frac{\partial V(X_{t+1}, P_{t+1})}{\partial X_{t+1}} \frac{d X_{t+1}}{d X_t} +
$$

$$
\frac{\partial V(X_{t+1}, P_{t+1})}{\partial P_{t+1}} \frac{d P_{t+1}}{d X_t} + \textcolor{coral}{\frac{\partial V(X_{t+1}, P_{t+1})}{\partial X_{t+1}} \cdot \frac{\partial X_{t+1}}{\partial C_t^*}} \cdot \frac{d C_t^*}{dX_t}
$$

The term in red is equal to zero according to equation $(2)$. Thus, it holds that

$$
\frac{d V(X_t, P_t)}{d X_t} = \beta R \frac{d V(X_{t+1}, P_{t+1})}{d X_{t+1}}
$$

$$
\frac{d V(X_t, P_t)}{d X_t} = u'(C_t) \quad \text{(given (1))}
$$

$$
\Rightarrow u'(C_t) = \beta R u'(C_{t+1})
$$
